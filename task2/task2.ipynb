```python
#問１

size = 5

for i in range(1, size + 1):
    print("*" * i)
```

```python
#問2

size = 5

for i in range(size):
    print(" " * i + "*")
```

```python
#問3

size = 5

for i in range(size):
    print(" " * (size - 1 - i) + "*")
```

```python
#問4

size = 5

for i in range(size):
    for j in range(size):
        if i == j or i + j == size - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()

```

```python
#問5

size = 5

for i in range(size):
    spaces = size - 1 - i
    stars = 2 * i + 1
    print(" " * spaces + "*" * stars)
```
